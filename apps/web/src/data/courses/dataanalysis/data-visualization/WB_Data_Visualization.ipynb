# Data Visualization with Plotly Express

This workbook covers interactive data visualization using **Plotly Express**, a high-level interface for creating rich, interactive charts with minimal code.

Plotly Express ships with several built-in datasets that we will use throughout this workbook — **no file downloads or CSV imports needed**. Each function returns a pandas DataFrame ready for plotting:
- `px.data.iris()` — Fisher's Iris flower measurements (150 rows)
- `px.data.tips()` — restaurant tipping data (244 rows)
- `px.data.gapminder()` — country-level economic and demographic indicators over time (1,704 rows)
- `px.data.stocks()` — normalized daily stock prices for 6 tech companies (100 rows)
- `px.data.experiment()` — simulated experimental results (100 rows)

### The Gapminder Dataset

The [Gapminder](https://www.gapminder.org/) dataset is the primary dataset used in this workbook. It was popularized by **Hans Rosling** in his famous TED talks and is ideal for practicing scatter plots, line charts, choropleth maps, and animations.

It contains **country-level indicators** for 142 countries, spanning from 1952 to 2007 (sampled every 5 years):

| Column | Description |
|---|---|
| `country` | Country name (142 countries) |
| `continent` | Africa, Americas, Asia, Europe, Oceania |
| `year` | Year of observation (1952–2007, every 5 years) |
| `lifeExp` | Life expectancy at birth (years) |
| `pop` | Total population |
| `gdpPercap` | GDP per capita (US$, inflation-adjusted) |
| `iso_alpha` | ISO 3166-1 alpha-3 country code (e.g. `USA`, `BRA`) |
| `iso_num` | ISO 3166-1 numeric country code |

**Original source:** https://www.gapminder.org/data/

**Documentation:** https://plotly.com/python/plotly-express/

## 2. Scatter Plots

Scatter plots show the relationship between two numeric variables. Each point represents one observation.

````python
fig = px.scatter(iris, x="sepal_width", y="sepal_length",
                 title="Iris: Sepal Width vs Sepal Length")
fig.show()
````

### Adding Color, Size, and Hover Data

We can encode additional dimensions using `color`, `size`, and `hover_data`.

````python
fig = px.scatter(
    iris,
    x="sepal_width",
    y="sepal_length",
    color="species",
    size="petal_length",
    hover_data=["petal_width"],
    title="Iris: Multi-dimensional Scatter Plot",
)
fig.show()
````

## 3. Line Charts

Line charts are ideal for visualizing trends over continuous data, especially time series.

````python
canada = gapminder.query("country == 'Canada'")
fig = px.line(canada, x="year", y="lifeExp",
              title="Life Expectancy in Canada")
fig.show()
````

### Comparing Multiple Lines

````python
countries = gapminder.query("country in ['Canada', 'Brazil', 'Japan', 'Nigeria']")
fig = px.line(countries, x="year", y="gdpPercap",
              color="country",
              title="GDP per Capita Over Time")
fig.show()
````

## 4. Bar Charts

Bar charts compare quantities across categories.

````python
fig = px.bar(tips, x="day", y="total_bill",
             color="sex", barmode="group",
             title="Total Bill by Day and Gender")
fig.show()
````

### Stacked Bar Chart

````python
fig = px.bar(tips, x="day", y="total_bill",
             color="sex", barmode="stack",
             title="Total Bill by Day (Stacked)")
fig.show()
````

### Horizontal Bar Chart

````python
americas_2007 = gapminder.query("year == 2007 and continent == 'Americas'")
fig = px.bar(americas_2007, y="country", x="gdpPercap",
             orientation="h",
             title="GDP per Capita in the Americas (2007)")
fig.update_layout(yaxis={"categoryorder": "total ascending"})
fig.show()
````

## 5. Histograms & Distributions

Histograms show the distribution of a single numeric variable.

````python
fig = px.histogram(tips, x="total_bill", nbins=30,
                   title="Distribution of Total Bill")
fig.show()
````

### Overlaid Histograms

````python
fig = px.histogram(tips, x="total_bill", nbins=30,
                   color="sex", barmode="overlay",
                   opacity=0.7,
                   title="Total Bill Distribution by Gender")
fig.show()
````

### Box Plots

````python
fig = px.box(tips, x="day", y="total_bill",
             color="smoker",
             title="Bill Distribution by Day and Smoking Status")
fig.show()
````

### Violin Plots

````python
fig = px.violin(tips, x="day", y="total_bill",
                color="sex", box=True,
                title="Bill Distribution (Violin + Box)")
fig.show()
````

## 6. Pie Charts & Hierarchical Charts

Part-to-whole relationships: how individual parts contribute to a total.

````python
continent_pop = gapminder.query("year == 2007").groupby("continent")["pop"].sum().reset_index()
fig = px.pie(continent_pop, values="pop", names="continent",
             title="World Population by Continent (2007)")
fig.show()
````

### Sunburst Chart

````python
df_2007 = gapminder.query("year == 2007")
fig = px.sunburst(df_2007, path=["continent", "country"],
                  values="pop",
                  title="World Population Hierarchy (2007)")
fig.show()
````

### Treemap

````python
fig = px.treemap(df_2007, path=["continent", "country"],
                 values="pop", color="lifeExp",
                 color_continuous_scale="RdYlGn",
                 title="Population & Life Expectancy (2007)")
fig.show()
````

## 7. Heatmaps & Correlation

Heatmaps visualize matrices of values using color intensity.

````python
iris_numeric = iris.select_dtypes(include="number")
corr = iris_numeric.corr()

fig = px.imshow(corr, text_auto=".2f",
                color_continuous_scale="RdBu_r",
                title="Iris Feature Correlation Matrix")
fig.show()
````

## 8. Scatter Matrix (Pair Plot)

A scatter matrix shows pairwise relationships between multiple numeric variables at once.

````python
fig = px.scatter_matrix(
    iris,
    dimensions=["sepal_width", "sepal_length", "petal_width", "petal_length"],
    color="species",
    title="Iris: Scatter Matrix",
)
fig.update_traces(diagonal_visible=False)
fig.show()
````

## 9. Maps & Geospatial Visualization

Plotly Express supports choropleth maps and scatter plots on geographic projections.

````python
df_2007 = gapminder.query("year == 2007")
fig = px.choropleth(df_2007, locations="iso_alpha",
                    color="gdpPercap",
                    hover_name="country",
                    color_continuous_scale="Viridis",
                    title="GDP per Capita by Country (2007)")
fig.show()
````

## 10. Animation

Plotly Express can animate charts over a variable (typically time) using `animation_frame`.

````python
fig = px.scatter(
    gapminder,
    x="gdpPercap",
    y="lifeExp",
    size="pop",
    color="continent",
    hover_name="country",
    animation_frame="year",
    animation_group="country",
    log_x=True,
    size_max=60,
    range_x=[100, 100000],
    range_y=[25, 90],
    title="Gapminder: GDP vs Life Expectancy (Animated)",
)
fig.show()
````

## 13. Customizing Appearance

Plotly Express figures can be customized using `update_layout()` and built-in templates.

````python
fig = px.scatter(iris, x="sepal_width", y="sepal_length",
                 color="species", size="petal_length")

fig.update_layout(
    template="plotly_dark",
    title="Iris Dataset (Dark Theme)",
    title_font_size=24,
    xaxis_title="Sepal Width (cm)",
    yaxis_title="Sepal Length (cm)",
    legend_title="Species",
)
fig.show()
````

Available templates: `plotly`, `plotly_white`, `plotly_dark`, `ggplot2`, `seaborn`, `simple_white`
